```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image_path = r"C:\Users\acer\OneDrive\Pictures\Camera Roll\Virat Kohli.jpg"

image = cv2.imread(image_path)

if image is None:
    raise FileNotFoundError("Image not found. Check the image path.")

print("Image loaded successfully!")
print("Image dimensions:", image.shape)
```

```python
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(rgb)
plt.title("Original Virat Kohli Image")
plt.axis("off")
plt.show()
```

```python
averaging = cv2.blur(image, (5, 5))

plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(averaging, cv2.COLOR_BGR2RGB))
plt.title("Averaging Filter")
plt.axis("off")
plt.show()
```

```python
kernel = np.array([
    [1, 2, 1],
    [2, 4, 2],
    [1, 2, 1]
], dtype=np.float32) / 16

weighted = cv2.filter2D(image, -1, kernel)

plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(weighted, cv2.COLOR_BGR2RGB))
plt.title("Weighted Averaging Filter")
plt.axis("off")
plt.show()
```

```python
gaussian = cv2.GaussianBlur(image, (5, 5), 0)

plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB))
plt.title("Gaussian Filter")
plt.axis("off")
plt.show()
```

```python
median = cv2.medianBlur(image, 5)

plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(median, cv2.COLOR_BGR2RGB))
plt.title("Median Filter")
plt.axis("off")
plt.show()
```

```python
laplacian_kernel = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
], dtype=np.float32)

sharpened = cv2.filter2D(image, -1, laplacian_kernel)

plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(sharpened, cv2.COLOR_BGR2RGB))
plt.title("Laplacian Sharpening")
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

laplacian = cv2.Laplacian(
    gray,
    cv2.CV_64F
)

laplacian_abs = cv2.convertScaleAbs(laplacian)

plt.figure(figsize=(6, 6))
plt.imshow(laplacian_abs, cmap="gray")
plt.title("Laplacian Operator - Grayscale")
plt.axis("off")
plt.show()
```

```python
images = [
    ("Original Image", rgb),
    ("Averaging Filter", cv2.cvtColor(averaging, cv2.COLOR_BGR2RGB)),
    ("Weighted Averaging", cv2.cvtColor(weighted, cv2.COLOR_BGR2RGB)),
    ("Gaussian Filter", cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB)),
    ("Median Filter", cv2.cvtColor(median, cv2.COLOR_BGR2RGB)),
    ("Laplacian Sharpening", cv2.cvtColor(sharpened, cv2.COLOR_BGR2RGB)),
    ("Grayscale Image", gray),
    ("Laplacian Operator", laplacian_abs)
]

plt.figure(figsize=(16, 12))

for i, (title, img) in enumerate(images, 1):
    plt.subplot(2, 4, i)

    if len(img.shape) == 2:
        plt.imshow(img, cmap="gray")
    else:
        plt.imshow(img)

    plt.title(title)
    plt.axis("off")

plt.tight_layout()
plt.show()
```
